Number Methods in Python

Round()

```python
num = 7.567
print(round(num, 2)) #Rounding the number to 2 decimal places
```

Format ()

```python
p = 3.14159
print(format(p,".2f"))
```

abs() Printing Absolute Value of a Number

```python
print(abs(-42))
```

Extracting Hex/Oct/Bin versions of a given number

```python
n = 255
print(hex(n))
print(oct(n))
print(bin(n))
```

int() and float()

```python
print(int(3.9))
print(float(3))
```

Complex Number

```python
z = complex(3, 4)

print(z)
print(z.real)
print(z.imag)
```

Div Mode: Get Quotient and Remainder at once

```python
print(divmod(10,3))
```
